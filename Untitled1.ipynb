```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

result = generator(
    "Artificial Intelligence",
    max_length=50,
    temperature=0.7,
    do_sample=True,
    top_k=50,
    top_p=0.95,
    num_return_sequences=1
)

print(result[0]["generated_text"])
```
