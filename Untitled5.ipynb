```python
import gradio as gr

def smart_home(temp):
    if temp > 26:
        return f"🌡️ Temperature: {temp}°C\n❄️ AC ON"
    elif temp < 22:
        return f"🌡️ Temperature: {temp}°C\n🔴 AC OFF"
    else:
        return f"🌡️ Temperature: {temp}°C\n✅ Comfortable - No Action"

gr.Interface(
    fn=smart_home,
    inputs=gr.Number(label="Enter Room Temperature (°C)"),
    outputs=gr.Textbox(label="Agent Decision"),
    title="🏠 Smart Home Temperature Control",
    description="Enter the room temperature to see the intelligent agent's decision."
).launch()
```
