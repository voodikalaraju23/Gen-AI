```python


temperature = float(input("Enter room temperature: "))

if temperature > 25:
    print("Temperature is high")
    print("AC: ON")
elif temperature < 20:
    print("Temperature is low")
    print("AC: OFF")
else:
    print("Temperature is comfortable")
    print("AC: OFF")
```
