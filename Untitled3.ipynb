```python
import gradio as gr

def smart_home_agent(temperature):
    if temperature > 26:
        return f"🌡️ Room Temperature: {temperature} °C\n\n❄️ Action: Air Conditioner ON"
    elif temperature < 22:
        return f"🌡️ Room Temperature: {temperature} °C\n\n🔴 Action: Air Conditioner OFF"
    else:
        return f"🌡️ Room Temperature: {temperature} °C\n\n✅ Comfortable Temperature - No Change"

interface = gr.Interface(
    fn=smart_home_agent,
    inputs=gr.Slider(10, 40, value=25, label="Room Temperature (°C)"),
    outputs=gr.Textbox(label="Agent Decision"),
    title="🏠 Smart Home Temperature Control Agent",
    description="Enter the room temperature and let the intelligent agent decide whether to switch the AC ON or OFF."
)

interface.launch()
```
